```python
import pandas as pd
```

```python
indiana_projections_df = pd.read_csv("../data/raw/indiana_projections.csv")
```

```python
indiana_projections_df
```

```python
# Range to check: 1 to 3999
expected_uids = set(range(1, 4000))
actual_uids = set(indiana_projections_df['uid'])

# Identify missing UIDs
missing_uids = sorted(list(expected_uids - actual_uids))

# Identify repeated UIDs
uid_counts = indiana_projections_df['uid'].value_counts()
repeated_uids = uid_counts[uid_counts > 1].index.tolist()

print(f"Missing UIDs (Total: {len(missing_uids)}): {missing_uids}")
print(f"Repeated UIDs (Total: {len(repeated_uids)}): {repeated_uids}")
```

We only need the frontal images so removing the other images i.e. Lateral Ones

```python
# 1. Keep only Frontal projections
indiana_projections_df = indiana_projections_df[indiana_projections_df['projection'] == 'Frontal']

# 2. Deduplicate UIDs (keep first occurrence for cases with multiple frontals)
indiana_projections_df = indiana_projections_df.drop_duplicates(subset=['uid'], keep='first')

print(f"Final count of unique frontal images: {len(indiana_projections_df)}")
```

```python
indiana_projections_df
```

```python
# Verification: Check if there are any remaining repeated UIDs
uid_counts = indiana_projections_df['uid'].value_counts()
repeated_uids = uid_counts[uid_counts > 1].index.tolist()
is_unique = indiana_projections_df['uid'].is_unique

print(f"Is UID unique? {is_unique}")
print(f"Repeated UIDs (Total: {len(repeated_uids)}): {repeated_uids}")
print(f"Total unique frontal images: {len(indiana_projections_df)}")
```

```python
# Cross-reference with reports to identify missing images
indiana_reports_df = pd.read_csv("../chest-x-ray-data/indiana_reports.csv")
report_uids = set(indiana_reports_df['uid'])
frontal_uids = set(indiana_projections_df['uid'])

missing_frontal = report_uids - frontal_uids
print(f"Total reports: {len(indiana_reports_df)}")
print(f"Reports with at least one frontal image: {len(frontal_uids & report_uids)}")
print(f"Reports missing frontal images: {len(missing_frontal)}")
```

```python
image_file_names = set(indiana_projections_df['filename'])
image_file_names
```

```python
import glob
import os
import shutil
```

```python
images = glob.glob(r'..\data\images\images_normalized\*.png')
destination_folder = r'..\data\processed_images'

# create folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)
```

Moving the lateral x-ray images into another folder

```python

for image in images:
   file_name = os.path.basename(image)
   if file_name not in image_file_names:
      destination_path = os.path.join(destination_folder, file_name)
      shutil.move(image, destination_path)
      print(f"Moved the file {file_name}")
```

Removing lateral images

```python
shutil.rmtree(f'../data/processed_images/')
```
